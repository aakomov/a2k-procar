```python
import boto3
import pandas as pd
from io import BytesIO

def read_car_data():
    s3_client = boto3.client(
        's3',
        endpoint_url='http://localhost:9090',
        aws_access_key_id='minio',
        aws_secret_access_key='minio123',
        verify=False
    )
    
    try:
        # Читаем файл
        #response = s3_client.get_object(Bucket='mlflow', Key='car_data_cleaned.parquet')
        response = s3_client.get_object(Bucket='mlflow', Key='part-00000-03df2316-ffcb-44b7-8018-afb9a13d66f7-c000.snappy.parquet')
        parquet_file = BytesIO(response['Body'].read())
        
        # Загружаем в DataFrame
        df = pd.read_parquet(parquet_file)
        
        print(f"✅ Успешно загружено! Размер данных: {df.shape}")
        print(f"📊 Колонки: {df.columns.tolist()}")
        
        return df
        
    except Exception as e:
        print(f"❌ Ошибка: {e}")
        return None

# Читаем данные
df = read_car_data()

if df is not None:
    print("\nПервые 5 строк:")
    print(df.head())
    
    print("\nБазовая информация:")
    print(df.info())
    
    print("\nСтатистика:")
    print(df.describe())
```
